Import Libraries

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import NearestNeighbors
import joblib
import os
```

Load Facebook Data (Adjusting Paths Correctly)

```python
# Load Facebook train edges
train_edges_fb = pd.read_csv('../data/facebook/train_edges.txt', sep=' ', header=None, names=['source', 'target'])

# Load Facebook test edges
test_edges_fb = pd.read_csv('../data/facebook/test_edges.txt', sep=' ', header=None, names=['source', 'target'])

# Load Facebook features
features_fb = pd.read_csv('../data/facebook/features.txt', sep=' ', header=None)
```

DATA CLEANING

```python
# Check Missing Values
print("Missing values:\n", features_fb.isnull().sum())

# Fill Missing Values (if any)
features_fb = features_fb.fillna(0)

# Validate User IDs
n_users = features_fb.shape[0]
max_user_train = train_edges_fb[['source', 'target']].max().max()
max_user_test = test_edges_fb[['source', 'target']].max().max()

print(f"Total users: {n_users}")
print(f"Max user in train: {max_user_train}")
print(f"Max user in test: {max_user_test}")

# Normalize Features
scaler_fb = StandardScaler()
features_fb_scaled = scaler_fb.fit_transform(features_fb)

print("✅ Facebook Data is cleaned and normalized!")
```
